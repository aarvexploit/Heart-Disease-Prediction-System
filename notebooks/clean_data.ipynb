```python
# Importing Modules and variables
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("../data/raw_data.csv")
cleaned_data = data.copy()
```

```python
# Setting Target as With Disease Yes - 1, No - 0
cleaned_data["target"] = (cleaned_data["num"] > 0).astype(int)
print(cleaned_data["target"].value_counts())
```

```python
# Removing Unwanted columns from data
cleaned_data = cleaned_data.drop(columns=["id", "dataset", "num", "ca", "thal"])
```

```python
# Replacing zero with NaN values
cleaned_data["chol"] = cleaned_data["chol"].replace(0, np.nan)
cleaned_data["trestbps"] = cleaned_data["trestbps"].replace(0, np.nan)
```

```python
# Filling Numeric Columns with Median
for col in ["trestbps", "chol", "thalch", "oldpeak"]:
    cleaned_data[col] = cleaned_data[col].fillna(cleaned_data[col].median())
```

```python
# Filling Boolean Columns with Mode
for col in ["fbs", "restecg", "exang"]:
    cleaned_data[col] = cleaned_data[col].fillna(cleaned_data[col].mode()[0])
```

```python
# Filling Slope Columns with Unknown
cleaned_data["slope"] = cleaned_data["slope"].fillna("unknown")
```

```python
cleaned_data.head()
```

```python
# Checking Left Missing values and other Operations
print("Missing values left:", cleaned_data.isnull().sum().sum())
print(cleaned_data.shape)
```

```python
# Checking Zeros left in Dataset
columns = ["trestbps", "chol", "thalch"]
print("Zero values :")
for col in columns:
    print(col, ":", (cleaned_data[col] == 0).sum())
```

```python
#Exporting Cleaned Data to CSV file
cleaned_data.to_csv("../data/cleaned_data.csv")
```
